```python
# Feature Engineering
```

```python
df.columns
```

```python
df["Temperature_Status"] = pd.cut(
    df["Temperature_C"],
    bins=[float("-inf"), 5.1, 9.88, 14.66, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df.head(2)
```

```python
df["Humidity_Status"] = pd.cut(
    df["Humidity_Percent"],
    bins=[float("-inf"), 56.7, 75.85, float("inf")],
    labels=["Low", "Normal", "High"]
)
```

```python
df.head(2)
```

```python
df["Shelf_Life_Remaining_Days"] = (
df["Shelf_Life_Days"] - (df["Package_Age_Hours"] / 24)
)
```

```python
df["Shelf_Life_Remaining_%"] = (
    (df["Shelf_Life_Remaining_Days"] / df["Shelf_Life_Days"]) * 100
).clip(0, 100)
df["Freshness_Index"] = (
    df["Shelf_Life_Remaining_%"]
    - (df["Temperature_C"] - 5).abs() * 3
    - (df["Humidity_Percent"] - 70).abs() * 0.3
).clip(0, 100)
```

```python
df["Sensor_Health_Score"] = (
    100
    - (df["Temperature_C"] - 5).abs() * 3
    - (df["Humidity_Percent"] - 70).abs() * 0.5
    - df["Oxygen_Concentration_Percent"] * 0.2
).clip(0, 100)
```

```python
df["Environmental_Stress_Score"] = (
    (df["Temperature_C"] - 5).abs() * 5
    + (df["Humidity_Percent"] - 70).abs() * 0.5
    + df["Vibration_g"] * 15
    + df["Light_Exposure_lux"] / 100
).clip(upper=100)
```

```python
df["Gas_Exposure_Score"] = (
    df["Oxygen_Concentration_Percent"]
    + df["Carbon_Dioxide_Percent"]
    + df["Volatile_Gas_ppm"]
)
df["Gas_Exposure_Level"] = pd.cut(
    df["Gas_Exposure_Score"],
    bins=[-float("inf"), 4, 8, float("inf")],
    labels=["Low", "Medium", "High"]
)
```

```python
df.head(2)
```

```python
df["Cold_Chain_Status"] = np.where(
    (df["Temperature_C"] >= 2) & (df["Temperature_C"] <= 8),
    "Maintained",
    "Broken"
)
df["Package_Quality_Status"] = np.select([
        df["Surface_Integrity_Status"] == "Contamination Defect",
        df["Surface_Integrity_Status"] == "Surface Deformation",
        (df["Surface_Integrity_Status"] == "Seal Defect") |
        (df["Seal_Pressure_kPa"] < 85) |
        (df["Vibration_g"] > 0.3)],[ "Poor","Average","Good"],
    default="Excellent")
```

```python
df["Digital_Twin_State"] = pd.to_numeric(
    df["Digital_Twin_State"],
    errors="coerce"
)
```

```python
df.head(3)
```

```python
df["Spoilage_Probability"] = np.select([(df["Digital_Twin_State"] >= 0.75) &((df["Gas_Exposure_Level"] == "High") |
  (df["Cold_Chain_Status"] == "Broken") | (df["Shelf_Life_Remaining_%"] < 40)),(df["Digital_Twin_State"] >= 0.45) &
  ((df["Gas_Exposure_Level"] == "Medium") |(df["Shelf_Life_Remaining_%"] < 70))],
    ["High", "Medium"],
    default="Low")
df["Compliance_Status"] = np.where(
    (df["Temperature_C"] >= 2) &
    (df["Temperature_C"] <= 8) &
    (df["Humidity_Percent"] >= 60) &
    (df["Humidity_Percent"] <= 80) &
    (df["Risk_Score"] < 0.5),
    "Safe",
    "Unsafe"
)
```

```python
df.head(3)
```

These derived metrics help identify high-risk packages early, improve storage and transportation decisions, and support data-driven quality control across the food supply chain.
